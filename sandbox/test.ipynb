```python
from datetime import datetime, timedelta

import numpy as np
import polars as pl
import time_stream as ts

from dritimeseriesprocessor.models.domain_models.processing_config import ProcessingMethodConfig
from dritimeseriesprocessor.operations.aggregation.aggregation_methods import Mean, DailyRad
```

```python
dates = [datetime(2023, 1, 1) + timedelta(days=i) for i in range(10)]
radiation = [0.0, 0.0, 5.1, 10.2, 2.0, 0.2, 0.0, 3.0, 1.6, 0.0]

df = pl.DataFrame({"time": dates, "radiation": radiation})
```

```python
tf = ts.TimeFrame(
    df=df,
    time_name="time", 
    resolution = "PT30M"
)
```

```python
tf.pad()
```

```python
def create_method_config(periodicity: ts.Period) -> ProcessingMethodConfig:
    """Create a test MethodConfig.

    Args:
        periodicity: Aggregation periodicity

    Returns:
        MethodConfig for testing
    """
    return ProcessingMethodConfig(
        method="test",
        params={"aggregation_period": periodicity},
    )
```

```python
config = create_method_config(ts.Period.of_days(1))
```

```python
daily_rad = DailyRad()
DailyRad.run(daily_rad, tf, config)
```
